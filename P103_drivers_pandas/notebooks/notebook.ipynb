```python
import matplotlib.pyplot as plt
import pandas as pd

drivers = pd.read_csv(
    "../data/drivers.csv",
    sep=",",
    thousands=None,
    decimal=".",
)

timesheet = pd.read_csv(
    "../data/timesheet.csv",
    sep=",",
    thousands=None,
    decimal=".",
)

summary = (
    timesheet.groupby("driverId", as_index=False)[["hours-logged", "miles-logged"]]
    .sum()
    .merge(drivers[["driverId", "name"]], on="driverId")
    .sort_values("driverId")
    .reset_index(drop=True)
)

summary.to_csv(
    "../submission/summary.csv",
    sep=",",
    header=True,
    index=False,
)

top10 = summary.sort_values("miles-logged", ascending=False).head(10)

plt.figure(figsize=(10, 6))
plt.barh(top10["name"], top10["miles-logged"])
plt.gca().invert_yaxis()
plt.tight_layout()
plt.savefig("../submission/top10_drivers.png", bbox_inches="tight")
plt.close()
```
